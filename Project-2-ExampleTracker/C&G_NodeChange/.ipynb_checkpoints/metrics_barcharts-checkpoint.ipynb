```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

base_path = Path(r"C:\Users\ranji\OneDrive\Documents\GitHub\research-helper\new")


def plot_from_csv(csv_name: str, title: str, y_label: str):
    df = pd.read_csv(base_path / csv_name)
    ax = df.plot(x="node", y=["CPLEX", "GA"], kind="bar", figsize=(8, 4))
    ax.set_title(title)
    ax.set_xlabel("Nodes")
    ax.set_ylabel(y_label)
    ax.legend(title="Method")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()
```

## Runtime

```python
plot_from_csv("runtime.csv", "Runtime vs Nodes", "Runtime")
```

## Distance

```python
plot_from_csv("distance.csv", "Distance vs Nodes", "Distance")
```

## Travel Time

```python
plot_from_csv("travelTime.csv", "Travel Time vs Nodes", "Travel Time")
```

## Energy Consumption

```python
plot_from_csv("energy_consumption.csv", "Energy Consumption vs Nodes", "Energy Consumption")
```

## Energy Charge (CS and ERS)

```python
df_cs = pd.read_csv(base_path / "energy_charge_CS.csv")
df_ers = pd.read_csv(base_path / "energy_charge_ERS.csv")

combined = pd.DataFrame({
    "node": df_cs["node"],
    "CPLEX_CS": df_cs["CPLEX"],
    "GA_CS": df_cs["GA"],
    "CPLEX_ERS": df_ers["CPLEX"],
    "GA_ERS": df_ers["GA"]
})

ax = combined.plot(
    x="node",
    y=["CPLEX_CS", "GA_CS", "CPLEX_ERS", "GA_ERS"],
    kind="bar",
    figsize=(10, 4)
)
ax.set_title("Energy Charge vs Nodes (CS and ERS)")
ax.set_xlabel("Nodes")
ax.set_ylabel("Energy Charge")
ax.legend(title="Method / Type")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
